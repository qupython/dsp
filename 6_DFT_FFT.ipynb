       <div>  numpy.fft.fftfreq برای به‌دست‌آوردن فرکانس‌های متناظر با هر نقطه از DFT استفاده می‌شود. </div>
  <div> np.fft.fftfreq is used to calculate the frequencies associated with each sample in the DFT. The function takes two arguments: the length of the signal and the sampling interval (which is equal to the time difference between adjacent samples).
      == timestep </div>
        
    
    
    
    </span>

```python
import numpy as np
import matplotlib.pyplot as plt

# Create a sinusoidal signal
t = np.linspace(0, 1, 100)
x = 1* np.sin(2 * np.pi * 10 * t)

noise = np.random.normal(0, 1, len(t))
x=x + 0.5*noise 

x2  = 2 * np.sin(2 * np.pi * 30 * t)
x=x+x2

# Compute the DFT using numpy.fft.fft
dft = np.fft.fft(x)

# Compute the power spectrum
power_spectrum = np.abs(dft)**2

# Plot the power spectrum
freqs = np.fft.fftfreq(len(x), t[1] - t[0])
plt.plot(freqs, power_spectrum)

plt.xlabel('Frequency (Hz)')
plt.ylabel('Power')
plt.show()
```

```python

import matplotlib.pyplot as plt
from scipy.io import wavfile

# Load the audio file
sampling_freq, audio_signal = wavfile.read('speech.wav')
print("(sampling_freq =" ,sampling_freq)
print(audio_signal)
# Plot the audio signal
plt.plot(audio_signal);plt.xlabel('Sample');plt.ylabel('Amplitude')
plt.show()
```

```python
dft = np.fft.fft(audio_signal)
power_spectrum = np.abs(dft)**2

freqs = np.fft.fftfreq(len(audio_signal),1/sampling_freq)
plt.plot(freqs, power_spectrum)

plt.xlabel('Frequency (Hz)')
plt.ylabel('Power')
plt.show()
```
